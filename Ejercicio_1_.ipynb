```python
multiplo_3 = lambda num: num %3==0
print (multiplo_3(9))
print (multiplo_3(10))
```

```python
num_cubo = lambda num : num **3
print (num_cubo (3))
```

```python
producto = lambda n1, n2 : n1* n2
print (producto (4,5))
```

```python
mayor = lambda n1, n2 : n1 if n1 > n2 else n2
print (mayor (9,4))
print (mayor(2,23))
```

```python
empiezacon_a = lambda palabra: palabra[0]== "A" or palabra[0]== "a"
print (empiezacon_a ("Amor"))
print (empiezacon_a("arco"))
print (empiezacon_a("cosa"))
```

```python
c_a_f = lambda celsius : (celsius * 9/5) +32
print (c_a_f(0))
print (c_a_f(10))
```

```python
funciones = [
    lambda numero: numero * 2,
    lambda numero: numero + 10,
    lambda numero: numero ** 2
]
numero = int(input("Ingrese un número: "))

i = 0

while i < len(funciones):
    print(funciones[i](numero))
    i = i + 1
```
