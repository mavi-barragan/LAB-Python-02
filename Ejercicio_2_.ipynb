```python
numeros = [1, 2, 3, 4, 5]
resultado = [(n, n**2) for n in numeros]
print(resultado)
```

```python
numeros = [2, 5, 10, 12, 15, 20]
cubos = [n ** 3 for n in numeros if n > 10]
print(cubos)
```

```python
numeros= [1,2,3,4,5,6]
pares= [n for n in numeros if n %2==0]
print(pares)
```

```python
numeros= [1,2,3,4,5,6]
resultado= ["Valor:" + str(n) for n in numeros]
print (resultado)
```

```python
grados= [0,10,20,30]
f= [(c* 9/5)+32 for c in grados]
print (f)
```

```python
palabras= ["carro", "yo", "gato", "peluche", "rosado"]
mas_5= [palabra for palabra in palabras if len(palabra) >5]
print (mas_5)
```

```python
palabras= ["amor","cielo","cancion","luz","barco"]
iniciales= [palabra[0] for palabra in palabras]
print (iniciales)
```

```python
numeros= [1,2,3,4,5,6,7,8]
p_o_i=["par" if n %2==0 else "impar" for n in numeros  ]
print(p_o_i)
```

```python
matriz= [[1,2,3], [4,5,6], [7,8,9]]
aplanada= [numero for fila in matriz for numero in fila]
print(aplanada)
```
