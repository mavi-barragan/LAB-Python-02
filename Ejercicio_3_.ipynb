```python
temp_c= [0,10,20,20]
def con_f(temp_c):
  return(temp_c * 9/5)+32
temp_f= list(map(con_f, temp_c))
print(temp_f)
```

```python
lista1= [1,2,3,4,5]
lista2= [3,4,5,6,7]
def suma(lista1, lista2):
  return (lista1 + lista2)
total= list(map(suma, lista1,lista2))
print (total)
```

```python
rad= [1,2,3,4]
def area (rad):
  return 3.1416 * rad ** 2
total= list (map(area,rad))
print (total)
```

```python
productos= {"pan": 1000, "leche": 2500, "café": 5000}
def aumentar(nombre):
    return (nombre, productos[nombre] * 1.10)
resultado = list(map(aumentar, productos))
print(resultado)
```

```python
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
def multiplicar(fila):
    return list(map(lambda numero: numero * 10, fila))
resultado = list(map(multiplicar, matriz))
print(resultado)
```
